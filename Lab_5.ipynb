```python
class Node:
    def __init__(self, state, val):
        self.state = state
        self.val = val
    def __str__(self):
        return f"State: {self.state}, Value: {self.val}"

landscape = {
    0: 2,
    1: 4,
    2: 7,
    3: 5,
    4: 3,
    5: 6,
    6: 9,
    7: 7,
    8: 4
}

def getneighbours(node):
    neighbours = []
    for position in [node.state - 1, node.state + 1]:
        if position in landscape:
            neighbours.append(Node(position, landscape[position]))
    return neighbours

def hillclimbing(start):
    current = Node(start, landscape[start])

    visited = []
    while True:
        print("Current:", current)
        visited.append(current.state)
        neighbours = getneighbours(current)
        if not neighbours:
            break

        bestneighbour = max(neighbours, key=lambda node: node.val)
        if bestneighbour.val <= current.val:
            print("Stopped: no better neighbours found")
            break

        current = bestneighbour

    return current, visited

solution, visited_states = hillclimbing(0)
print("\nVisited states:", visited_states)
print("Final result:", solution)
```

A student wants to divide study time between Programming and Mathematics. There are 10 hours available.

Here, p represents the hours allocated to Programming. Mathematics receives the remaining hours, so 10−p hours.

**Requirements:**

* Represent a schedule as a Node.
* Define the objective function using the benefit formula.
* Generate neighbors by moving one hour between subjects.
* Use hill climbing to maximize the benefit.
* Display the final allocation and benefit.

Expected result: The maximum occurs when Programming receives 6 hours and Mathematics receives 4 hours.

```python
class StudyNode:
    def __init__(self, p_hours):
        self.p_hours = p_hours
        self.m_hours = 10 - p_hours
        self.val = self.calculate_benefit()

    def calculate_benefit(self):
        prog_benefit = 12 * self.p_hours - (self.p_hours ** 2)
        math_benefit = 8 * self.m_hours - (self.m_hours ** 2)
        return prog_benefit + math_benefit

    def __str__(self):
        return f"Programming: {self.p_hours}h, Math: {self.m_hours}h | Benefit: {self.val}"

def get_study_neighbours(node):
    neighbours = []
    for next_p in [node.p_hours - 1, node.p_hours + 1]:
        if 0 <= next_p <= 10:
            neighbours.append(StudyNode(next_p))
    return neighbours

def hill_climbing_study(start_p):
    current = StudyNode(start_p)
    print("Starting hill climbing...")
    while True:
        print("Current Allocation:", current)
        neighbours = get_study_neighbours(current)
        if not neighbours:
            break
        best_neighbour = max(neighbours, key=lambda node: node.val)
        if best_neighbour.val <= current.val:
            print("Stopped: local maximum reached")
            break
        current = best_neighbour
    return current

optimal_allocation = hill_climbing_study(0)
print("\nOptimal Allocation Found:")
print(optimal_allocation)
```

Suppose, you work at an urban planning centre. You wish to create a new nuclear power plant.The problem with this situation is that the power plant has to stay as far from the houses as possible to avoid harmful radioactive radiations. What is the final location you could find using hill climbing alogrithm.






```python
class LocationNode:
    def __init__(self, x):
        self.x = x
        self.val = self.calculate_objective()

    def calculate_objective(self):
        house_positions = [2, 5, 12, 18]
        return min(abs(self.x - h) for h in house_positions)

    def __str__(self):
        return f"Location x: {self.x} | Minimum Distance to any house: {self.val}"

def get_location_neighbours(node):
    neighbours = []
    for next_x in [node.x - 1, node.x + 1]:
        if 0 <= next_x <= 20:
            neighbours.append(LocationNode(next_x))
    return neighbours

def hill_climbing_plant(start_x):
    current = LocationNode(start_x)
    print(f"Starting hill climbing from x = {start_x}...")
    while True:
        print("Current:", current)
        neighbours = get_location_neighbours(current)
        if not neighbours:
            break
        best_neighbour = max(neighbours, key=lambda node: node.val)
        if best_neighbour.val <= current.val:
            print("Stopped: local maximum reached")
            break
        current = best_neighbour
    return current

optimal_location = hill_climbing_plant(7)
print("\nOptimal Power Plant Location Found:")
print(optimal_location)
```
