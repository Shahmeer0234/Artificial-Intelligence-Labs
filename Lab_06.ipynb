```python
import random
import math

'''landscape = {
    0: 2,
    1: 4,
    2: 7,
    3: 5,
    4: 3,
    5: 6,
    6: 9,
    7: 7,
    8: 4
}'''

landscape = {
0: 8,
1: 5,
2: 2,
3: 6,
4: 9,
5: 4,
6: 1,
7: 3,
8: 7,
}


def get_neighbors(x):
    #Previously we just did
    #return [x-1 , x+1]
    # WHY are we checking if the value is in the landscape
    neighbors = []
    if x - 1 in landscape:
        neighbors.append(x - 1)

    if x + 1 in landscape:
        neighbors.append(x + 1)

    return neighbors
```

```python
def simulated_annealing(start, temperature, cooling_rate):

    current = start
    best = current

    while temperature > 0.1:
        neighbors = get_neighbors(current)

        # Select a random neighboring state
        next_state = random.choice(neighbors)

        current_value = landscape[current]
        next_value = landscape[next_state]

        # Difference for a maximization problem
        delta = next_value - current_value

        # Accept every better move
        if delta >= 0:
            current = next_state

        # Sometimes accept a worse move
        else:
            probability = math.exp(delta / temperature)

            if random.random() < probability:
                current = next_state

        # Keep track of the best solution ever found
        if landscape[current] > landscape[best]:
            best = current

        # Cool down
        temperature *= cooling_rate

    return best


solution = simulated_annealing(start=2,temperature=10,cooling_rate=0.95)

print("Best state:", solution)
print("Best value:", landscape[solution])
```

```python
import random
import math

def objective_function(p):
    # Quadratic function banayi hai jiska peak p = 6 par ho aur max value 36 mile
    return -((p - 6) ** 2) + 36

def get_neighbors_study(p):
    neighbors = []
    if p - 1 >= 0:
        neighbors.append(p - 1)
    if p + 1 <= 10:
        neighbors.append(p + 1)
    return neighbors

def simulated_annealing_study(start, temperature, cooling_rate):
    current = start
    best = current

    while temperature > 0.1:
        neighbors = get_neighbors_study(current)
        next_state = random.choice(neighbors)

        current_value = objective_function(current)
        next_value = objective_function(next_state)

        delta = next_value - current_value

        if delta >= 0:
            current = next_state
        else:
            probability = math.exp(delta / temperature)
            if random.random() < probability:
                current = next_state

        if objective_function(current) > objective_function(best):
            best = current

        temperature *= cooling_rate

    return best

solution = simulated_annealing_study(start=2, temperature=10, cooling_rate=0.95)

print("Best Programming hours (p):", solution)
print("Best Mathematics hours:", 10 - solution)
print("Best Satisfaction Score:", objective_function(solution))
```

**Exercise:**\
Suppose, you work at an urban planning centre. You wish to create a new nuclear power plant.The problem with this situation is that the power plant has to stay as far from the houses as possible to avoid harmful radioactive radiations. What is the final location you could find using simulated annealing alogrithm.






```python
import random
import math

# Maan lete hain houses position 5 par hain, toh distance maximize karni hai
def distance_objective(x):
    return abs(x - 5)

def get_neighbors_plant(x):
    neighbors = []
    if x - 1 >= 0:
        neighbors.append(x - 1)
    if x + 1 <= 10:
        neighbors.append(x + 1)
    return neighbors

def simulated_annealing_plant(start, temperature, cooling_rate):
    current = start
    best = current

    while temperature > 0.1:
        neighbors = get_neighbors_plant(current)
        next_state = random.choice(neighbors)

        current_value = distance_objective(current)
        next_value = distance_objective(next_state)

        delta = next_value - current_value

        if delta >= 0:
            current = next_state
        else:
            probability = math.exp(delta / temperature)
            if random.random() < probability:
                current = next_state

        if distance_objective(current) > distance_objective(best):
            best = current

        temperature *= cooling_rate

    return best

solution_plant = simulated_annealing_plant(start=2, temperature=10, cooling_rate=0.95)

print("Best Power Plant Location:", solution_plant)
print("Maximum Distance Value:", distance_objective(solution_plant))
```
