#### Knowledge Based Agents

**Lab Outcomes**
* Understand how an AI agent can make intelligent decisions using facts, knowledge, and logical rules instead of simply following fixed instructions.

**Exercise:**
The IT department receives complaints that computers cannot print documents.

**Knowledge:**
- If the printer is switched off, infer printer_unavailable.
- If the printer is switched on but the computer cannot detect it, infer printer_connection_problem.
- If the printer is available but has no paper, infer paper_unavailable.
  
**Requirements:**
1. Collect the required observations.
2. Add the new rules to the knowledge base.
3. Add corresponding recommendations.
4. Test all three situations.

```python
rules = {
    "printer_unavailable": ["printer_off"],
    "printer_connection_problem": ["printer_on", "computer_cannot_detect"],
    "paper_unavailable": ["printer_available", "no_paper"],
}

# Recommendations
recommendations = {
    "printer_unavailable": "Turn on the printer and check power cable.",
    "printer_connection_problem": "Check cable/Wi-Fi or reinstall printer driver.",
    "paper_unavailable": "Load paper in the printer tray.",
}

test_cases = [
    ["printer_off"],
    ["printer_on", "computer_cannot_detect"],
    ["printer_available", "no_paper"],
]

for case in test_cases:
    facts = case.copy()

    # Inference engine
    for conclusion in rules:
        conditions = rules[conclusion]
        if all(c in facts for c in conditions):
            facts.append(conclusion)

    print("Facts:", facts)
    for problem in recommendations:
        if problem in facts:
            print("Recommendation:", recommendations[problem])
    print()
```

**Exercise:**
Develop an intelligent irrigation agent that decides whether crops need watering.

**Constraints:**
- Soil moisture below 30% = dry soil.
- Soil moisture above 70% = wet soil.
- Dry soil AND high temperature = urgent watering.
- Wet soil = do not water.

**Requirements:**
1. Input soil moisture.
2. Input temperature.
3. Define the rules.
4. Use the inference engine to derive conclusions.
5. Display the recommended action.

```python
# Input soil moisture and temperature
moisture = float(input("Enter soil moisture (%): "))
temperature = float(input("Enter temperature (C): "))

facts = []

if moisture < 30:
    facts.append("dry_soil")
if moisture > 70:
    facts.append("wet_soil")
if temperature > 30:
    facts.append("high_temperature")

rules = {
    "urgent_watering": ["dry_soil", "high_temperature"],
    "do_not_water": ["wet_soil"],
    "normal_watering": ["dry_soil"],
}

# Recommendations
recommendations = {
    "urgent_watering": "Urgent watering needed.",
    "do_not_water": "Do not water.",
    "normal_watering": "Water the crops.",
}

# Inference engine
for conclusion in rules:
    conditions = rules[conclusion]
    if all(c in facts for c in conditions):
        facts.append(conclusion)

print("Facts:", facts)
for state in recommendations:
    if state in facts:
        print("Action:", recommendations[state])
```
