**Exercise:**
Write a program that displays the following menu:
Geometry Calculator
1. Calculate the Area of a Circle
2. Calculate the Area of a Rectangle
3. Calculate the Area of a Triangle
4. Quit
   
Enter your choice (1–4):
* If the user enters 1, the program should ask for the radius of the circle and then display its area. Use 3.14159 for PI
* If the user enters 2, the program should ask for the length and width of the rectangle, and then display the rectangle’s area.
* If the user enters 3, the program should ask for the length of the triangle’s base and height, and then display its area.
* If the user enters 4, the program should end.

Use the match-case statement for handling the user choices. Also handle the invalid inputs


```python
#Python While Loops
choice = 1
while choice:
    print("1. Calculate the Area of Circle\n2. Calculate Area of Rectangle\n3. Calculate Area of Triangle\n4. Quit")
    choice = int(input("Enter your Choice: "))

    match choice:
        case 1:
            rad = float(input("\nEnter Radius of Circle: "))
            print("Area of Circle is: ", 2 * 3.14159 * rad)
        case 2:
            length = float(input("Enter Length: "))
            width = float(input("Enter Width: "))
            print("\nArea of Rectangle is: ", length * width)
        case 3:
            base = float(input("Enter Base: "))
            height = float(input("Enter Height: "))
            print("\nArea of Triangle: ", 1/2 * base * height)
        case 4:
            print("Thanks for your hospitality!")
            break
        case _:
            print("Unknown Error Occured")
```

e.g newBook -> new_book

```python
#Functions Creation, Calling, Return
def Vowels():
    input_user = str(input("Enter a Word: "))
    vowels = ['a', 'e', 'i', 'o', 'u']
    count = 0
    for char in input_user.lower():
        if char in vowels:
            count += 1
    print("Number of vowels: ", count)

Vowels()

camel = str(input("Enter variable name in camelCase: "))
snake = ""
for char in camel:
    if char.isupper():
        snake += "_" + char.lower()
    else:
        snake += char

print(snake)
# Default Parameters, Keyword Parameters, Return
```

Write a function that determines if a given year is a leap year (divisible by 4, but not by 100 unless also divisible by 400).

**Exercise**
Count the frequency of words in the text below and store it in a dictionary where each key is the word counted.

```python
text = "python is easy and python is powerful"

words = text.split()
freq = {}

for word in words:
    if word in freq:
        freq[word] += 1
    else:
        freq[word] = 1

print(freq)
```
